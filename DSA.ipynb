## **Dynamic Programming**

### **Tips:**
- Better give naming conventions meaningful instead of s1, i, coz this makes the code readable and easy to understand.
- Always/ try do recursion from **top-down** (n --> 0)
  - Express problem in terms of index
  - Try out all possibilities ( means Recursion ) according to the problem statement
  - Then finalize solution as per requirement in terms of sum, max, min, count, ... etc
- Memoization follows recursion, so same process but `dp` array makes big significance, while computing compared to recursion.
  - Declare dp array, update it accordingly. Check and use it after base case if value is available ( i != -1 )
- Tabulation, Space-Optimized Tabulation opposite to recursion i.e., **bottom-up** (0 --> n)

### **How to Solve?**
- Always try to derive your logic from 0th or middle or last element and then slowly increase/decrease, coz by this u'll get some clarity and also here is wehre u'll get ur most important part of recursion i.e., **base case**.
- First to know what exactly is your solve(n)? like what exactly u'll get from f(1) means --> [0 - 1 idx], the same way f(3) --> [0 ... 3] and so on, is that fulfilling ur requirement. (which is asked in the problem)
- That's how u should compute and build-up solution like piece/ step by step with clarity

##### **2. LC-62: [Unique Paths](https://leetcode.com/problems/unique-paths/) ⭐**

Given the two integers m and n, return the number of possible unique paths from top-left corner to bottom-right corner by taking move either down or right at any point in time.

**Constraints:**
- 1 <= m, n <= 100

````python
def uniquePaths(m: int, n: int) -> int:
  # '''
  # Space-Optimized Tabulation - T.C = O(), S.C = O()
  dpu = [1] * m
  dpl = [1] * n
  # dpl[0] = dpu[0] = 1

  for i in range(1, m):
    for j in range(1, n):
      left = dpl[j-1]
      up = dpu[i-1]
      dpl[j] = left + up
  return dpl[j]
  # '''

  '''
  # Tabulation - T.C = O(), S.C = O(N*M)
  dp = [[0] * (n) for _ in range(m)]

  for i in range(m):
    dp[i][0] = 1
  for j in range(n):
    dp[0][j] = 1
  # print(dp)

  for i in range(1, m):
    for j in range(1, n):
      left = dp[i][j-1]
      up = dp[i-1][j]
      dp[i][j] = left + up
    # print(dp)
  return dp[i][j]
  # '''

  '''
  # Memoization - T.C = O(N*M), S.C = O(N+M) + O(N*M)
  dp = [[-1] * n for _ in range(m)]

  def solve(i, j):
    if i == 0 and j == 0: return 1
    if i < 0 or j < 0: return 0
    if dp[i][j] != -1: return dp[i][j]

    left = solve(i, j-1)
    up = solve(i-1, j)
    dp[i][j] = left + up
    # print(dp)
    return dp[i][j]

  return solve(m-1, n-1)
  # '''

  '''
  # Recursion - T.C = O(2^(M*N)), S.C = O(M+N)
  def solve(i, j):
    if i == 0 and j == 0: return 1
    if i < 0 or j < 0: return 0

    left = solve(i, j-1)
    up = solve(i-1, j)
    return left + up

  return solve(m-1, n-1)
  # '''

m, n = 3, 7 # map(int, input().split()) # m - row, n - cols
print(uniquePaths(m, n)) # 28
````

If we observe carefully then here we can try out all possibilities - which means we can use **Recursion**

**1. Express problem in terms of index**
- So here express in terms of idx means `figuring out base cases`.

- As I told you in recursion, memoization always do from **top-to-bottom** (n --> 0).

- Now if we start from $(m-1, n-1)$ then anyhow we gonna reach $(0, 0)$.

- So if you remember in recursion **count sum match** pblm, there we  returned $1$ only if condition is matched/ satisfied, else $0$.

- So here our satisfied condition means reaching to $(0, 0)$ from $(m-1, n-1)$, meanwhile in this process we may sometimes cross out of boundary which is the case of **$i < 0$ or $j < 0$** then we should return $0$ immediatley to avoid unexpected values/ behaviour.

**2. Try out all possibilities/ Explore/ Do all the stuffs**
- As per pblm statement given that can move to right or down at any point in time.

- So for this condition stating/ satisfying we should write a logic:
```python
  left = solve(i, j-1)
  up = solve(i-1, j)
```

**3. Finalize solution as per requirement in terms of sum, max, min, count, ... etc**
- As per pblm stmnt final answer should return total number of possible unique paths, which is **count**.
- So for that we return up + left simply:
```python
  return left + up
```



##### **✅1. GfG: [Ninja's Training](https://www.geeksforgeeks.org/problems/geeks-training/1) ⭐⭐**

Each day he must perform exactly one of three activities:
- Running - mat[i][0], Fighting - mat[i][1], Learning - mat[i][2]

To improve all his skills, Geek cannot perform the same activity on two consecutive days. Return the maximum total merit points.

**Constraints:**

- 1 ≤ mat.size() ≤ $10^5$
- 1 ≤ mat[i][j] ≤ 100
- mat[i].size() = 3

````python
def maximumPoints(mat: list[list[int]]) -> int:
  n = len(mat)

  # '''
  # Space-Optimized Tabulation - T.C = O(N*3*3), S.C = O(N)
  if n == 1: return max(mat[0])
  dp = mat[0]

  for day in range(1, n):
    temp = [0] * 3
    for cur in range(3):
      for last in range(3):
        if cur != last:
          points = mat[day][cur] + dp[last]
          temp[cur] = max(temp[cur], points)
    dp = temp
  return max(dp)

  """ # below code can be written as above

  for day in range(1, n):
    dp1 = [0] * 3
    for cur in range(3):
      maxPoints = 0
      for last in range(3):
        if cur != last:
          points = mat[day][cur] + dp[last]
          maxPoints = max(maxPoints, points)
      dp1[cur] = maxPoints # below & this gives same o/p but time costs slightly
    dp = dp1
    # print(*dp)
  return max(dp)
  """
  # '''

  '''
  # Tabulation - T.C = O(N*3*3), S.C = O(3*N)
  if n == 1: return max(mat[0])
  dp = [[0] * 3 for _ in range(n)]
  dp[0] = mat[0]

  for day in range(1, n):
    for cur in range(3):
      maxPoints = 0
      for last in range(3):
        if cur != last:
          points = mat[day][cur] + dp[day-1][last]
          maxPoints = max(maxPoints, points)
          dp[day][cur] = maxPoints
    # print(*dp)
  return max(dp[day])
  # '''

  '''
  # Memoization - T.C = O(4*N), S.C = O(4*N) + O(N) [Recursive Stack Space]
  dp = [[-1] * 4 for _ in range(n)]

  def solve(day, last):
    maxPoints = 0

    if day == 0:
      for i in range(3):
        if i != last:
          maxPoints = max(maxPoints, mat[day][i])
      return maxPoints

    if dp[day][last] != -1: return dp[day][last]

    for i in range(3):
      if i != last:
        points = mat[day][i] + solve(day-1, i)
        maxPoints = max(maxPoints, points)
    dp[day][last] = maxPoints
    # print(*dp)
    return dp[day][last]

  return solve(n-1, 3)
  # '''

  '''
  # Recursion - T.C = O(3^N), S.C = O(N)
  def solve(day, last):
    maxPoint = 0

    if day == 0:
      for i in range(3): # "i" is here current/ today's task
        if i != last:
          maxPoint = max(maxPoint, mat[day][i])
      return maxPoint

    for i in range(3): # here also "i" is same
      if i != last:
        point = mat[day][i] + solve(day-1, i) # testing out current day
        maxPoint = max(maxPoint, point) # getting maxPoints for that curr day
    return maxPoint

  return solve(n-1, 3)
  # '''

  """
   1. Express in terms of index -> So here `day` is our index.

   2. Try out stuff -> In this process we got to know that without knowing
      what is `last` task we cannot perform next day task coz to avoid
      the same task doing 2 consecutive days.

   3. `3` indicates that no task was done previously, so by this we can try out
      all 3 tasks on first day, then afterwards we skip that last day task.
      Even if there is only 1 day given as n (n=1)

   Note: Consecutive days - day after day
  """

n = 4 # int(input()) # Not given, I added for `loop` purpose to make it easy
# mat = [list(map(int, input().split())) for _ in range(n)]
mat = [[2, 5, 3], [6, 4, 1], [3, 7, 2], [4, 16, 9]]
print(maximumPoints(mat)) # 29
# Explanation: [5 + 6 + 2 + 16 = 29]
````

##### **✅6. LC-213: [House Robber II](https://leetcode.com/problems/house-robber-ii/)** ⭐

 Extension for the House Robber I problem. Here all houses at this place are **arranged in a circle**. That means the first house is the neighbor of the last one.

**Constraints:**

- 1 <= nums.length <= 100
- 0 <= nums[i] <= 1000

````python
def rob(nums: list[int]) -> int:
  n = len(nums)
  if n == 1: return nums[0] # For Single element purpose

  # '''
  # Space-Optimized Tablation - T.C = O(N), S.C = O(1)
  def solve(i, j):
    prev2 = prev = 0

    for i in range(i, j):
      s1 = nums[i] + prev2
      prev2, prev = prev, max(s1, prev)
    return prev

  return max(solve(0, n-1), solve(1, n))
  # '''

  '''
  # Tablation - T.C = O(N), S.C = O(N)
  def solve(i, j):
    dp = [0] * n

    for i in range(i, j):
      s1 = nums[i] + dp[i-2]
      s2 = dp[i-1]
      dp[i] = max(s1, s2)
    return dp[i]

  return max(solve(0, n-1), solve(1, n))
  # '''

  '''
  # Memoization - T.C = O(N), S.C = O(N) + O(N) [Recursive Stack Space]
  dp = [-1] * n

  def solve(i, j):
    if i < j: return 0
    if dp[i] != -1: return dp[i]

    s1 = nums[i] + solve(i-2, j)
    s2 = solve(i-1, j)
    dp[i] = max(s1, s2)
    return dp[i]

  s1 = solve(n-2, 0)
  dp = [-1] * n
  return max(s1, solve(n-1, 1))
  # '''

  """
  - Remember in this problem, we have to place `dp` array carefully instead of
    outside of loop coz for odd number length we are calculating 2 times, so
    dp must be refreshed instead of using same old dp.

  - For this purpose we are placing dp inside `solve` block for Tabulation and
    for Memoization 1 more time declared just before the 2nd time calling.
  """

  '''
  # Recursion - T.C = O(2^N), S.C = O(N)
  def solve(i, j):
    if i < j: return 0

    s1 = nums[i] + solve(i-2, j)
    s2 = 0 + solve(i-1, j)
    return max(s1, s2)

  return max(solve(n-2, 0), solve(n-1, 1))
  # '''

nums = [2,3,2,1,2,1] # list(map(int, input(),split()))
print(rob(nums)) # 3
# Explanation: [3], Coz [2,2] both are neighbours which alerts the police.
````

##### **✅5. LC-198: [Maximum Sum of Non-Adjacent Elements](https://leetcode.com/problems/house-robber/) | [House Robber I](https://leetcode.com/problems/house-robber/)**
Given an integer array nums representing the amount of money of each house, return the maximum amount of money you can rob tonight without alerting the police, by not robbing two adjacent houses due to security alert systems.

**Constraints:**

- 1 <= nums.length <= 100
- 0 <= nums[i] <= 400

````python
def rob(nums: list[int]) -> int:
  n = len(nums)

  # '''
  # Space-Optimized Tabulation - T.C = O(N), S.C = O(1)
  prev1, prev = 0, nums[0]

  for i in range(1, n):
    s1 = nums[i] + prev1
    s2 = prev
    prev1, prev = prev, max(s1, s2)
  return prev
  # '''

  """
  - In above code prev1 = 0 just following i < 0 case as initially i-2 --> -1
  - Also, we can simply write above and below code as follows:
      for i in range(1, n):
        s1 = nums[i] + prev1
        prev1, prev = prev, max(s1, prev)
      return prev
  """

  '''
  # Tabulation - T.C = O(N), S.C = O(N)
  dp = [0] * n
  dp[0] = nums[0]

  for i in range(1, n):
    s1 = nums[i] + dp[i-2]
    s2 = dp[i-1]
    dp[i] = max(s1, s2)
  return dp[i]
  # '''

  '''
  # Memoization - T.C = O(N), S.C = O(N) + O(N) [Recursive Stack Space]
  dp = [-1] * n

  def solve(i):
    if i == 0: return nums[0]
    if i < 0: return 0
    if dp[i] != -1: return dp[i]

    s1 = nums[i] + solve(i-2)
    s2 = solve(i-1)

    dp[i] = max(s1, s2)
    print(*dp)
    return dp[i]
  return solve(n-1)
  # '''

  '''
  # Recursion - T.C = O(2^N), S.C = O(N) + O(N) [Recursive Stack Space]
  def solve(i):
    if i == 0: return nums[0]
    if i == 1: return max(nums[0], nums[1])

    """
    - if i < 0: return 0 # above base case (i == 1) and this works same
    - Purpose is to avoid negative idx like in s1 = solve(1-2) -> -1
    - But if we use i==1 then it won't go to s1 case, direct gives max element
    - Where in the case of i==0 it executes & calls solve(i-2) and then returns
      value 0. Check memoization there I written i < 0 case, same output.
    """

    s1 = nums[i] + solve(i-2)
    s2 = 0 + solve(i-1) ( not pick )
    return max(s1, s2)

  return solve(n-1)
  # '''

nums = [2,7,9,3,1] # list(map(int, input().split()))
print(rob(nums)) # 12 - Explanation: [2 + 9 + 1 = 12]
````

##### **✅4. TUF-714: [Frog Jump with K distances](https://takeuforward.org/practice/dsa/frog-jump-with-k-distances) ⭐**

Extension for the Frog Jump problem, here frog can jump from the ith step to any step in the range [i + 1, i + k]

Return the minimum amount of energy required by the frog to go from the 0th step to the (n-1)th step.

**Constraints:**
- 1 <= n <= $10^4$
- 1 <= k <= 10
- 0 <= heights[i] <= $10^4$

**Note:** Plus Subscription needed to submit above pblm. So what u can do this - hard code 'k' value as **2** and submit it in GfG portal at Frog Jump problem for verification purpose ( if needed ).

````python
def frogJump(heights: list[int], k: int) -> int:
  n = len(heights)

  # '''
  # Space-Optimized Tabuation - T.C = O(K*N), S.C = O(K)
  if n-1 == 0: return 0
  dp = [0] * k

  for i in range(1, n):
    s1 = float('inf')
    for j in range(1, k+1):
      if i >= j:
        x = abs(heights[i] - heights[i-j]) + dp[(i-j)%k]
        s1 = min(s1, x)
    dp[i%k] = s1

  return dp[i%k]
  # '''

  '''
  # Tabulation - T.C = O(K*N), S.C = O(N)
  if n-1 == 0: return 0
  dp = [0] * n

  for i in range(1, n):
    s1 = float('inf')
    for j in range(1, k+1):
      if i >= j:
        x = abs(heights[i] - heights[i-j]) + dp[i-j]
        s1 = min(s1, x)
    dp[i] = s1

  return dp[i]
  # '''

  '''
  # Memoization - T.C = O(K*N), S.C = O(N) + O(N) [Recursive Stack Space]
  dp = [-1] * n # declaring the dp array

  def solve(i):
    if i == 0: return 0
    if dp[i] != -1: return dp[i]

    s1 = float('inf')
    for j in range(1, k+1):
      if i >= j:
        x = abs(heights[i] - heights[i-j]) + solve(i-j)
        s1 = min(s1, x)
    # print(*dp)
    dp[i] = s1 # updating to dp array

    return dp[i] # as 'i' is curr and dp[i] is also curr, so returning
  return solve(n-1)
  # '''

  '''
  # Recursion - T.C = O(K^N), S.C = O(N)
  def solve(i):
    if i == 0: return 0

    s1 = float('inf')
    for j in range(1,k+1):
      if i >= j:
        x = abs(heights[i] - heights[i-j]) + solve(i-j)
        s1 = min(s1, x)

    return s1
  return solve(n-1)
  # '''

heights = [15, 4, 1, 14, 15] # list(map(int, input().split()))
k = 3 # int(input())
print(frogJump(heights, k)) # 2
# Explanation: [(15 -> 14 = 1), (14 -> 15 = 1) => 1 + 1 = 2]
````

##### **✅3. GfG: [Frog Jump](https://www.geeksforgeeks.org/problems/geek-jump/1)**

From any stair i, the frog has two options: it can either jump to the (i+1)th stair or the (i+2)th stair. The cost of a jump is the absolute difference in height between the two stairs.

Determine the minimum total cost required for the frog to reach the last stair.

**Constraints:**
- 1 ≤ height.size() ≤ $10^5$
- 0 ≤ height[i] ≤ $10^4$

````python
def minCost(height: list[int]) -> int:
  n = len(height)

  # '''
  # Space-Optimized Tabulation - T.C = O(N), S.C = O(1)
  if n-1 == 0: return 0

  prior2 = prior1 = 0

  for i in range(1, n):
    s1 = abs(height[i] - height[i-1]) + prior1

    s2 = float('inf')
    if i > 1:
      s2 = abs(height[i] - height[i-2]) + prior2

    prior2, prior1 = prior1, min(s1, s2)
  return prior1
  # '''

  '''
  # Tabulation - T.C = O(N), S.C = O(N)
  if n-1 == 0: return 0

  dp = [0] * n

  for i in range(1, n):
    s1 = abs(height[i] - height[i-1]) + dp[i-1]

    s2 = float('inf')
    if i > 1:
      s2 = abs(height[i] - height[i-2]) + dp[i-2]

    dp[i] = min(s1, s2)
  return dp[i]
  # '''

  '''
  # Memoization - T.C = O(N), S.C = O(N) + O(N) [Recursive Stack Space]
  dp = [-1] * n
  def minJump(i):
    if i == 0: return 0
    if dp[i] != -1: return dp[i]

    s1 = abs(height[i] - height[i-1]) + minJump(i-1)

    s2 = float('inf')
    if i > 1:
      s2 = abs(height[i] - height[i-2]) + minJump(i-2)

    dp[i] = min(s1, s2)
    return dp[i]

  return minJump(n-1)
  # '''

  '''
  # Recursion - T.C = O(2^N), S.C = O(N)
  def minJump(i):
    if i == 0: return 0

    s1 = abs(height[i] - height[i-1]) + minJump(i-1)

    s2 = float('inf')
    if i > 1:
      s2 = abs(height[i] - height[i-2]) + minJump(i-2)

    return min(s1, s2)

  return minJump(n-1)
  # '''

height = [10, 30, 40, 20, 50] # list(map(int, input().split()))
print(minCost(height)) # 40
# Explanation: [(30 -> 60 = 30), (60 -> 60 = 0), (60 -> 50 = 10) => 30 + 0 + 10 = 40]
````

##### **✅2. LC-70: [Climbing Stairs](https://leetcode.com/problems/climbing-stairs/)**

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

**Constraints:**

- 1 <= n <= 45

````python
def climbStairs(n: int) -> int:
  # '''
  # Space-Optimized Tabulation - T.C = O(N), S.C = O(1)
  prev1 = prev = 1

  for i in range(2, n+1):
    cur = prev1 + prev
    prev1, prev = prev, cur

  return prev
  # '''

  '''
  # Tabulation - T.C = O(N), S.C = O(N)
  if n <= 1: return 1

  dp = [0] * (n+1)
  dp[0] = dp[1] = 1

  for i in range(2, n+1):
    dp[i] = dp[i-1] + dp[i-2]

  return dp[n]
  # '''

  '''
  # Memoization - T.C = O(N), S.C = O(N) + O(N) [Recursive Stack Space]
  dp = [-1] * (n+1)

  def solve(n):
    if n <= 1:
      dp[n] = 1
      return dp[n]

    if dp[n] != -1: return dp[n]
    dp[n] = solve(n-1) + solve(n-2)
    return dp[n]

  return solve(n)
  # '''

  '''
  # Recursion - T.C = O(2^N), S.C = O(N)
  if n <= 1: return 1
  return climbStairs(n-1) + climbStairs(n-2)
  # '''

n = 4 # int(input())
print(climbStairs(n)) # 5 - Explanation: [1111, 211, 121, 112, 22]
````

##### **✅1. LC-509: [Fibonacci Number](https://leetcode.com/problems/fibonacci-number/description/)**

The Fibonacci numbers, commonly denoted F(n) form a sequence, called the Fibonacci sequence, such that each number is the sum of the two preceding ones, starting from 0 and 1. That is,

F(0) = 0, F(1) = 1

F(n) = F(n - 1) + F(n - 2), for n > 1. Given n, calculate F(n).

**Constraints:**

- 0 <= n <= 30

````python
def fib(n: int) -> int:
  # '''
  # Space-Optimized Tabulation -T.C = O(N), S.C = O(1)
  if n <= 1: return n

  prev1, prev = 0, 1
  for i in range(2, n+1):
    cur = prev + prev1
    prev1, prev = prev, cur
  return prev
  # '''

  '''
  # Tabulation - T.C = O(N), S.C = O(N)
  if n <= 1: return n

  dp = [0] * (n+1)
  dp[1] = 1

  for i in range(2, n+1):
    dp[i] = dp[i-1] + dp[i-2]
  return dp[n]
  # '''

  '''
  # Memoization - T.C = O(N), S.C = O(N) + O(N) [Recursive Stack Space]
  dp = [-1] * (n+1)

  def solve(n):
    if n <= 1: return n
    if dp[n] != -1: return dp[n]
    dp[n] = solve(n-1) + solve(n-2)
    return dp[n]

  return solve(n)
  # '''

  '''
  # Recursion - T.C = O(2^N), S.C = O(N)
  if n <= 1: return n
  return fib(n-1) + fib(n-2)
  # '''

n = 6 # int(input())
print(fib(n)) # 8 - Explanation: [0 1 1 2 3 5 8]
````

##### **✅Quick Sort ⭐**
It follows logic of pick any element as pivot then compare rest of elements and place accordingly ( left to the pivot if element is smaller and vice-versa to the right ), then pivot will be placed at respective sorted place, then same will be iterated until sorted.  


````python
def partition(low, high):
  i, j, pivot = low, high, arr[low]

  while i < j:
    while i < high and arr[i] <= pivot: # "<=" - To accept equal element or else TLE err will occur
      i += 1

    while j > low and arr[j] > pivot:
      j -= 1

    if i < j:
      arr[i], arr[j] = arr[j], arr[i] # mistake - while swapping didn't twisted values

  arr[low], arr[j] = arr[j], pivot
  return j # mistake - didn't returned value


def quickSort(low, high):
  if low < high:
    pIdx = partition(low, high)
    quickSort(low, pIdx-1) # mistake - instead of quickSort called partition
    quickSort(pIdx+1, high) # mistake - instead of quickSort called partition

arr = [3, 1, 6, 12, 4, 2, 1, 9, 11]
quickSort(0, len(arr)-1)
print(arr)
````

##### **✅Merge Sort ⭐**
It follows logic of divide ( split ) until single element by using mid [ (low+high)/2 ] then compare both sides and merge.

**Polished:** Operates by partitioning an array into smaller subarrays, individually sorting these subarrays, and subsequently combining the sorted subarrays to construct the ultimately sorted array.

````python
def compareAndMerge(low, mid, high):
  i, j, temp = low, mid+1, []

  while i <= mid and j <= high:
    if arr[i] < arr[j]:
      temp.append(arr[i])
      i += 1
    else:
      temp.append(arr[j])
      j += 1

  while i <= mid:
    temp.append(arr[i])
    i += 1

  while j <= high:
    temp.append(arr[j])
    j += 1

  for k in range(low, high+1):
    arr[k] = temp[k-low]


def divide(low, high):
  if low == high: return

  mid = (low + high) // 2

  divide(low, mid)
  divide(mid+1, high)
  compareAndMerge(low, mid, high)

arr = [3, 1, 6, 12, 4, 2, 1, 9, 11]
divide(0, len(arr)-1)
print(arr)
````

##### **✅Insertion Sort**

It divides a list into two parts: a part that's already sorted and an unsorted part.

It then repeatedly selects an item from the unsorted part and places it in its appropriate position within the sorted part

This continues until the entire list is in the correct order

````python
def insertionSort(n):
  for i in range(1, n):
    while i > 0 and arr[i-1] > arr[i]:
      arr[i-1], arr[i] = arr[i], arr[i-1]
      i -= 1

arr = [30, 50, 40, 20, 10, 20]
insertionSort(len(arr))
print(arr)
````

##### **✅Bubble Sort**

A basic sorting method for a list of items, where it repeatedly compares and swaps adjacent items to move the larger ones to the end of the list until the entire list is sorted

````python
def bubbleSort(n):
  for i in range(n-1, 0, -1):
    isSorted = 1
    for j in range(i):
      if arr[j] > arr[j+1]: # mistake - given arr[i] instead of arr[j+1]
        arr[j], arr[j+1] = arr[j+1], arr[j] # mistake - due to above swapped with arr[i]
        isSorted = 0
    if isSorted:
      break

arr = [22, 1, 13, 5, 4, 12, 38, 1]
bubbleSort(len(arr))
print(arr)
````

##### **✅Selection Sort**

It finds the smallest item and puts it at the front until the whole list is sorted in ascending order.

````python
def selectionSort(n):
  for i in range(n):
    for j in range(i+1, n):
      if arr[i] > arr[j]:
        arr[i], arr[j] = arr[j], arr[i]

arr = [4, 12, 5, 16, 7, 24, 5, 9]
selectionSort(len(arr))
print(arr)
````
